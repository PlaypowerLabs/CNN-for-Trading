```python
import yfinance as yf
import pandas as pd
import os

# Create a folder to store individual datasets
os.makedirs("nifty50_datasets", exist_ok=True)

nifty_50_stocks = [
    'RELIANCE.NS', 'HDFCBANK.NS', 'ICICIBANK.NS', 'INFY.NS', 'TCS.NS',
    'HINDUNILVR.NS', 'ITC.NS', 'LT.NS', 'KOTAKBANK.NS', 'SBIN.NS',
    'BAJFINANCE.NS', 'BHARTIARTL.NS', 'AXISBANK.NS', 'ASIANPAINT.NS', 'MARUTI.NS',
    'HCLTECH.NS', 'TITAN.NS', 'SUNPHARMA.NS', 'HDFC.NS', 'ULTRACEMCO.NS',
    'M&M.NS', 'POWERGRID.NS', 'NTPC.NS', 'ADANIENT.NS', 'INDUSINDBK.NS',
    'NESTLEIND.NS', 'TATASTEEL.NS', 'WIPRO.NS', 'BAJAJFINSV.NS', 'DIVISLAB.NS',
    'DRREDDY.NS', 'CIPLA.NS', 'GRASIM.NS', 'HEROMOTOCO.NS', 'JSWSTEEL.NS',
    'BPCL.NS', 'BRITANNIA.NS', 'APOLLOHOSP.NS', 'COALINDIA.NS', 'EICHERMOT.NS',
    'ADANIPORTS.NS', 'HINDALCO.NS', 'UPL.NS', 'TATAMOTORS.NS', 'TATACONSUM.NS',
    'ONGC.NS', 'TECHM.NS', 'BAJAJ-AUTO.NS', 'SBILIFE.NS', 'SHREECEM.NS',
    'DMART.NS', 'ICICIPRULI.NS', 'PIDILITIND.NS', 'DLF.NS', 'VEDL.NS',
    'BANDHANBNK.NS', 'GODREJCP.NS', 'DABUR.NS', 'HDFCLIFE.NS', 'MCDOWELL-N.NS',
    'SIEMENS.NS', 'HAVELLS.NS', 'PGHH.NS', 'TRENT.NS', 'GLAND.NS',
    'IDBI.NS', 'BERGEPAINT.NS', 'COLPAL.NS', 'AUROPHARMA.NS', 'BOSCHLTD.NS',
    'ACC.NS', 'LTI.NS', 'INDIGO.NS', 'BIOCON.NS', 'TORNTPHARM.NS',
    'NAUKRI.NS', 'INDUSTOWER.NS', 'BEL.NS', 'TATAPOWER.NS', 'NMDC.NS',
    'PETRONET.NS', 'GAIL.NS', 'HDFCAMC.NS', 'PIIND.NS', 'ADANIGREEN.NS',
    'MOTHERSON.NS', 'ETERNAL.NS', 'LICI.NS', 'TVSMOTOR.NS', 'GODREJPROP.NS',
    'SRF.NS', 'CHOLAFIN.NS', 'TORNTPOWER.NS', 'BALKRISIND.NS', 'JINDALSTEL.NS',
    'LAURUSLABS.NS', 'MUTHOOTFIN.NS', 'ICICIGI.NS', 'ABB.NS', 'POLYCAB.NS',
    'MFSL.NS', 'CONCOR.NS', 'AUBANK.NS', 'SBICARD.NS', 'IGL.NS',
    'APLLTD.NS', 'WHIRLPOOL.NS', 'CROMPTON.NS', 'IRCTC.NS', 'IEX.NS'
]

# Step 1: Download and save individual datasets
for ticker in nifty_50_stocks:
    print(f"Downloading {ticker}...")
    data = yf.download(ticker, start="2024-09-01", end="2025-09-12")
    data["Ticker"] = ticker
    data.reset_index(inplace=True)

    # Save individual CSV
    filename = f"nifty50_datasets/{ticker.replace('.', '_')}.csv"
    data.to_csv(filename, index=False)

print("✅ All individual datasets saved in 'nifty50_datasets/' folder")

# Step 2: Club all datasets into one
all_data = []

for ticker in nifty_50_stocks:
    filename = f"nifty50_datasets/{ticker.replace('.', '_')}.csv"
    df = pd.read_csv(filename)
    all_data.append(df)

final_df = pd.concat(all_data, ignore_index=True)

# Save final combined CSV
final_df.to_csv("nifty50_combined.csv", index=False)

print("✅ Final combined dataset saved as 'nifty50_combined.csv'")
```
